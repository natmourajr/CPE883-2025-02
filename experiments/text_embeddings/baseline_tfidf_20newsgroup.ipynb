```python
from sklearn.feature_extraction.text import TfidfVectorizer
from collections import defaultdict
from time import time

from sklearn import metrics
import numpy as np
```

```python
from datasets import concatenate_datasets
import hf_dataset_loader 
```

```python
dl = hf_dataset_loader.TwentyNewsgroupsDataset()
```

```python
dataset = concatenate_datasets(dl.dataset.values())
```

```python
texts = list(dataset['text'])
labels = np.array(list(dataset['label']))
```

```python
from collections import Counter
```

```python
counter = Counter(labels.tolist())
counter.most_common()
len(counter)
```

```python
vectorizer = TfidfVectorizer(
    max_df=0.5,
    min_df=5,
    max_features=5000,
    stop_words="english",
)
t0 = time()
X_tfidf = vectorizer.fit_transform(texts).toarray()

print(f"vectorization done in {time() - t0:.3f} s")
print(f"n_samples: {X_tfidf.shape[0]}, n_features: {X_tfidf.shape[1]}")
```

```python
from text_embeddings.clustering import bootstrap_clustering_metrics
```

```python
resp = bootstrap_clustering_metrics(X_tfidf, labels)
```
